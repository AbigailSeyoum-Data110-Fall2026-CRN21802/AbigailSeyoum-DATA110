```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Anscombe's Quartet:
x = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y1 = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
y2 = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
y3 = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]
x4 = [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8]
y4 = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
```

```python
plt.scatter(x, y1)
```

```python
# Calculate the slope (m) and intercept (b) of the line using np.polyfit
m, b = np.polyfit(x, y1, 1)

# Create the regression line
regression_line = m * np.array(x) + b

# Plot the data points and regression line
plt.scatter(x, y1)
plt.plot(x, regression_line,color='red')
plt.xlabel('x')
plt.ylabel('y1')
plt.title('Linear Regression for y1')
plt.show()
```

```python
m2, b2 = np.polyfit(x, y2, 1)
regression_line2 = m2 * np.array(x) + b2

plt.scatter(x, y2)
plt.plot(x, regression_line2, color='red')
plt.xlabel('x')
plt.ylabel('y2')
plt.title('Linear Regression for y2')
plt.show()
m3, b3 = np.polyfit(x, y3, 1)
regression_line3 = m3 * np.array(x) + b3
```

```python
plt.scatter(x, y3)
plt.plot(x, regression_line3, color='red')
plt.xlabel('x')
plt.ylabel('y3')
plt.title('Linear Regression for y3')
plt.show()
m4, b4 = np.polyfit(x4, y4, 1)
regression_line4 = m4 * np.array(x4) + b4
```

```python
plt.scatter(x4, y4)
plt.plot(x4, regression_line4, color='red')
plt.xlabel('x4')
plt.ylabel('y4')
plt.title('Linear Regression for y4')
plt.show()
```

Reflection:
1. How do the datasets visually differ despite having similar summary statistics?

The datasets all have a positive linear regression, but the correlation is stronger with x3,y3 than x1,y1, x2,y2, and x4,y4 (despite the outlier).

2. How did the outlier in the x4, y4 dataset affect the regression line compared to the other datasets?

The outlier in the x4,y4 dataset (19, 12.50) makes the correlation weaker.

3. Why is it important to visualize data in addition to calculating summary statistics?

Visualizing data can help one come to a conclusion about positive or negative correlation, and also identify any outliers in a dataset.
