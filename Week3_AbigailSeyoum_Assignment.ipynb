```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    'Offense': [
        'Aggravated Assault', 'Burglary', 'Larceny',
        'Motor Vehicle Theft', 'Murder', 'Rape (revised)', 'Robbery'
    ],
    'Offenses_2023': [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5],
    'Change_2022_2023': [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]
}

df = pd.DataFrame(data)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

colors1 = '#1f77b4'
bars1 = ax1.barh(df['Offense'], df['Offenses_2023'], color=colors1, alpha=0.85)
ax1.set_title('Offenses per 100,000 People (2023)', fontsize=12, fontweight='bold', pad=10)
ax1.set_xlabel('Offenses per 100,000 People', fontsize=10)
ax1.grid(True, linestyle='--', alpha=0.5, axis='x')
ax1.invert_yaxis()  # Highest/first on top

for bar in bars1:
    width = bar.get_width()
    ax1.text(width + 15, bar.get_y() + bar.get_height()/2, f'{width:,.1f}',
             va='center', ha='left', fontsize=9, fontweight='bold')

colors2 = ['#5cb85c' if x < 0 else '#d9534f' for x in df['Change_2022_2023']]
bars2 = ax2.barh(df['Offense'], df['Change_2022_2023'], color=colors2, alpha=0.85)
ax2.set_title('Change in Offenses per 100k (%) (2022–2023)', fontsize=12, fontweight='bold', pad=10)
ax2.set_xlabel('Percentage Change (%)', fontsize=10)
ax2.axvline(0, color='black', linewidth=1)
ax2.grid(True, linestyle='--', alpha=0.5, axis='x')
ax2.invert_yaxis()

# Add values on bars
for bar in bars2:
    width = bar.get_width()
    offset = 0.5 if width >= 0 else -0.5
    ha = 'left' if width >= 0 else 'right'
    ax2.text(width + offset, bar.get_y() + bar.get_height()/2, f'{width:+.1f}%',
             va='center', ha=ha, fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('crime_bar_graph.png', bbox_inches='tight', dpi=300)
```

```python
# Dataset
companies = ['Company A', 'Company B', 'Company C']
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

# X-axis positioning for grouped bars
x = np.arange(len(companies))
width = 0.25  # Width of each bar

# Create figure
fig, ax = plt.subplots(figsize=(9, 6))

# Plot bars for each product with offsets
rects1 = ax.bar(
    x - width,
    product1,
    width,
    label='Product 1',
    color='#1f77b4',
    edgecolor='black',
    alpha=0.85,
)
rects2 = ax.bar(
    x,
    product2,
    width,
    label='Product 2',
    color='#ff7f0e',
    edgecolor='black',
    alpha=0.85,
)
rects3 = ax.bar(
    x + width,
    product3,
    width,
    label='Product 3',
    color='#2ca02c',
    edgecolor='black',
    alpha=0.85,
)

# Labels, Title, and Legend
ax.set_ylabel('Sales Units', fontsize=12)
ax.set_xlabel('Companies', fontsize=12)
ax.set_title(
    'Product Sales Comparison by Company',
    fontsize=14,
    fontweight='bold',
    pad=12,
)
ax.set_xticks(x)
ax.set_xticklabels(companies, fontsize=11)
ax.legend(title='Products', fontsize=10)
ax.set_ylim(0, 42)
ax.grid(True, linestyle='--', alpha=0.5, axis='y')

# Data value labels on top of each bar
ax.bar_label(rects1, padding=3, fontweight='bold')
ax.bar_label(rects2, padding=3, fontweight='bold')
ax.bar_label(rects3, padding=3, fontweight='bold')

plt.tight_layout()
plt.show()
```
